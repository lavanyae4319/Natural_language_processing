```python
import pandas as pd
import re

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report
```

```python

data = {
    "News": [
        "Scientists discover new treatment for cancer",
        "Government announces new education policy",
        "World Cup tournament will start next month",
        "New technology helps doctors detect diseases",
        "University releases official examination results",
        "Aliens have landed and taken over the city",
        "Drinking magical water can cure every disease",
        "Celebrity claims the moon will disappear tomorrow",
        "Secret machine can make people invisible",
        "Government will give unlimited money to everyone",
        "Researchers publish a new study on climate change",
        "Bank announces new digital payment service",
        "Ghosts are controlling all mobile phones",
        "Eating one fruit can make you live forever",
        "Scientists develop a new renewable energy system"
    ],

    "Label": [
        "Real",
        "Real",
        "Real",
        "Real",
        "Real",
        "Fake",
        "Fake",
        "Fake",
        "Fake",
        "Fake",
        "Real",
        "Real",
        "Fake",
        "Fake",
        "Real"
    ]
}

df = pd.DataFrame(data)
```

```python
print("NEWS DATASET:")
print(df.head())
```

```python
def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text


df["Cleaned_News"] = df["News"].apply(preprocess)

print("\nCLEANED NEWS:")
print(df[["News", "Cleaned_News", "Label"]].head())
```

```python
vectorizer = TfidfVectorizer(
    stop_words="english"
)

X = vectorizer.fit_transform(
    df["Cleaned_News"]
)

y = df["Label"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42
)
```

```python
model = MultinomialNB()
model.fit(
    X_train,
    y_train
)
```

```python
y_pred = model.predict(X_test)

print("\nPREDICTIONS:")
print(y_pred)
```

```python
accuracy = accuracy_score(
    y_test,
    y_pred
)

print("\nMODEL ACCURACY:")
print(round(accuracy * 100, 2), "%")

print("\nCLASSIFICATION REPORT:")
print(
    classification_report(
        y_test,
        y_pred
    )
)
```

```python
new_news = [
    "Scientists develop a new medicine for heart disease",
    "Aliens will control the entire world tomorrow"
]

new_news_cleaned = [
    preprocess(news)
    for news in new_news
]

new_news_vector = vectorizer.transform(
    new_news_cleaned
)

prediction = model.predict(
    new_news_vector
)
```

```python

print("\nNEW NEWS PREDICTIONS:")

for news, result in zip(
    new_news,
    prediction
):
    print(news)
    print("Prediction:", result)
    print()
```
